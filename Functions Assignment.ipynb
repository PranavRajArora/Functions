```python
###Q. 1. What is the difference between a function and a method in Python?
#A.1. A function is a block of code that performs a specific task and is defined using the def keyword and is called independently whereas a  method is a function that is associated with an object and is called on that object and is defined in a class and takes "self' as the first parameter.
###
#function example
def greetings(name):
    return f"Hello, {name}!"

print(greetings("Raj"))
```

```python
#method example
class Person:
    def greetings(self, name):
        return f"Hello, {name}!"

p = Person()
print(p.greetings("Raj"))
```

```python
###Q.2  Explain the concept of function arguments and parameters in Python.
###A.2. Parameters are the variables listed in a function that specifies which data the function will receive. They act as placeholders for the actual values that will be used during the function's execution. Arguments are the actual values provided to a function when it is called. These values are assigned to the corresponding parameters so that the function can operate on them.
def greetings(name):  # 'name' is a parameter, it has not yet taken a value
    print(f"Hello, {name}!")
greetings("Raj") #here, Raj is the argument that is assigned to the parameter 'name'
```

```python
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

# Using the generator
for num in count_up_to(3):
    print(num)
```

```python
###Q.1 Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in
the list.
def sum_even_numbers(numbers):
    total = 0
    for num in numbers:
        if num % 2 == 0:
            total += num
    return total
my_list = [12, 5, 7, 8, 10]
print(sum_even_numbers(my_list))
```

```python
###Q.4. Write a Python function that checks if a given number is prime or not from 1 to 200.

def is_prime(n):
    if n <= 1:
        return False  # 1 and numbers less than 1 are not prime
    for i in range(2, int(n ** 0.5) + 1):  # Check divisibility from 2 to sqrt(n)
        if n % i == 0:
            return False  # n is divisible by i, so it's not prime
    return True

# Check prime numbers from 1 to 200
prime_numbers = [num for num in range(1, 201) if is_prime(num)]
print(prime_numbers)
```

```python
###Q.5  Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

def fibo_generator(terms):
    a, b = 0, 1
    for _ in range(terms):
        yield a
        a, b = b, a + b

# Example usage:
for num in fibo_generator(10):  # Generate first 10 Fibonacci numbers
    print(num)
```

```python
###Q.6 Write a generator function in Python that yields the powers of 2 up to a given exponent.

def powers_of_two(exponent):
    for i in range(exponent + 1):  # Includes the given exponent
        yield 2 ** i

# Example usage:
for i in powers_of_two(5):  # Generate powers of 2 up to 2^5
    print(i)
```

```python
###Q.9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

celsius_temps = [0, 20, 30, 37, 100]
fahrenheit_temps = list(map(lambda c: c * 9/5 + 32, celsius_temps))
print(fahrenheit_temps)
```

```python
###Q.10. Create a Python program that uses `filter()` to remove all the vowels from a given string.

def remove_vowels(input_str):
    vowels = 'aeiouAEIOU'
    result = ''.join(filter(lambda char: char not in vowels, input_str))
    return result
text = "Hello, World!"
no_vowels = remove_vowels(text)
print(no_vowels)
```
